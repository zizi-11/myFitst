```python
import matplotlib.pyplot as plt
```

```python
numeric_cols = ['Critic_Score', 'User_Score', 'Global_Sales', 
                'NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']
```

```python
corr = df[numeric_cols].corr()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap between Sales and Review Scores')
plt.show()
```

```python
genre_stats = df.groupby('Genre').agg({
    'Global_Sales': 'sum',
    'Critic_Score': 'mean'
}).reset_index()

fig, ax1 = plt.subplots(figsize=(10,6))

ax1.bar(genre_stats['Genre'], genre_stats['Global_Sales'], color='skyblue', label='Global Sales')
ax1.set_xlabel('Genre')
ax1.set_ylabel('Global Sales (millions)', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')
plt.xticks(rotation=45)

ax2 = ax1.twinx()
ax2.plot(genre_stats['Genre'], genre_stats['Critic_Score'], color='red', marker='o', label='Critic Score')
ax2.set_ylabel('Average Critic Score (0–100)', color='red')
ax2.tick_params(axis='y', labelcolor='red')

plt.title('Global Sales and Average Critic Score by Genre')
plt.show()
```

```python
df['Score_Diff'] = df['Critic_Score'] - (df['User_Score'] * 10)  # match scale (0–100)
genre_diff = df.groupby('Genre')['Score_Diff'].mean().sort_values(ascending=False)
```

```python
plt.figure(figsize=(10,6))
plt.axhline(0, color='black', linestyle='--', linewidth=1)
plt.title('Critic vs User Agreement by Genre')
plt.xticks(rotation=45)
sns.barplot(y=genre_diff.index, x=genre_diff.values, palette='coolwarm')
plt.xlabel('Average Difference (Critic - User)')
plt.ylabel('Genre')
plt.show()
```

إذا كانت النتيجة موجبة (+) → يعني أن النقاد أعجبوا باللعبة أكثر من اللاعبين.

إذا كانت سالبة (-) → يعني أن اللاعبين أحبوا اللعبة أكثر من النقاد.

إذا كانت قريبة من الصفر (0) → يعني اتفاق بين الطرفين تقريبًا.
