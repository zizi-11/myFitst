```python
df["Year_of_Release"]= df["Year_of_Release"].astype(float)
```

```python
imputer = SimpleImputer(strategy='mean')
df['Year_of_Release'] = imputer.fit_transform(df[['Year_of_Release']])
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["Year_of_Release"]= df["Year_of_Release"].astype(int)
```

```python
df["Year_of_Release"]= df["Year_of_Release"].astype(str)
```

```python
df['Genre'] = df['Genre'].fillna('Unknown')
df
```

```python
df.isnull().sum()
```

```python
df['Name'] = df['Name'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df['Publisher'] = df['Publisher'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
#from sklearn.impute import SimpleImputer
```

```python
#pip install scikit-learn
```

```python
df['User_Score'] = df['User_Score'].replace('tbd', pd.NA)
```

```python
df['User_Score'] = pd.to_numeric(df['User_Score'], errors='coerce')
```

```python
average_score = df['User_Score'].mean()
df['User_Score'] = df['User_Score'].fillna(average_score)
```

```python
df['User_Score'] = df['User_Score'].round(2)
```

```python
df.isnull().sum()
```

```python
df['User_Count'] = df['User_Count'].replace('tbd', pd.NA)
```

```python
df['User_Count'] = pd.to_numeric(df['User_Count'], errors='coerce')
```

```python
average_count = df['User_Count'].mean()
df['User_Count'] = df['User_Count'].fillna(average_count)
```

```python
df.isnull().sum()
```

```python
df['User_Count'] = df['User_Count'].astype(int)
```

```python
df['Developer'] = df['Developer'].fillna('Unknown')
```

```python
most_frequent_rating = df['Rating'].mode()[0]
df['Rating'] = df['Rating'].fillna(most_frequent_rating)
```

```python
df.isnull().sum()
```

```python
df['Rating'] = df['Rating'].fillna(method='ffill')
```

```python
df.isnull().sum()
```

```python
imputer = SimpleImputer(strategy='mean')
df['Critic_Score'] = imputer.fit_transform(df[['Critic_Score']])
df['Critic_Score'] = df['Critic_Score'].round(2)
```

```python
df['Critic_Count'] = imputer.fit_transform(df[['Critic_Count']])
df['Critic_Count'] = df['Critic_Count'].round(0)
df
```
